```python
def preprocessing_tail(path):
    #make folder
    try:
        os.mkdir(path + 'Tail//')
    except FileExistsError:
        pass
    
    #gather raw data
    with os.scandir(path) as entries:
        for entry in entries:
            if entry.is_dir():
                with os.scandir(entry.path) as babyentries:
                    for babyentry in babyentries:
                        if 'behavior_log' in babyentry.name and 'tail' in entry.name:
                            bh_path = babyentry.path
                        elif 'metadata' in babyentry.name:
                            metadata_path = babyentry.path
    tail_df = pd.read_csv(bh_path, sep=';', index_col=0)
    
    file = open(metadata_path)
    metadata = file.read()
    start_time = metadata[metadata.find('t_protocol_start": "'):].split('"')[2]
    start_time = dt.strptime(start_time, '%Y-%m-%dT%H:%M:%S.%f')
    file.close()
    
    #normalize for time
    running_time = []
    for t in tail_df.t:
        time = start_time + timedelta(seconds = t)
        running_time = running_time + [time]
    tail_df.loc[:, 'real_time'] = running_time
    
    #save file
    tail_df.to_hdf(path + 'Tail//tail_df.hdf', key = 'tail', mode = 'w')
    
def preprocessing_eye(path):
    #make folder
    try:
        os.mkdir(path + 'Eye//')
    except FileExistsError:
        pass
    
    #gather raw data
    with os.scandir(path) as entries:
        for entry in entries:
            if entry.is_dir() and 'eye' in entry.name:
                with os.scandir(entry.path) as babyentries:
                    for babyentry in babyentries:
                        if 'behavior_log' in babyentry.name:
                            bh_path = babyentry.path
                        elif 'metadata' in babyentry.name:
                            metadata_path = babyentry.path
    eye_df = pd.read_csv(bh_path, sep=';', index_col=0)
    
    file = open(metadata_path)
    metadata = file.read()
    start_time = metadata[metadata.find('t_protocol_start": "'):].split('"')[2]
    start_time = dt.strptime(start_time, '%Y-%m-%dT%H:%M:%S.%f')
    file.close()
    
    #normalize for time
    running_time = []
    for t in eye_df.t:
        time = start_time + timedelta(seconds = t)
        running_time = running_time + [time]
    eye_df.loc[:, 'real_time'] = running_time
    
    #save file
    eye_df.to_hdf(path + 'Eye//eye_df.hdf', key = 'tail', mode = 'w')

def simple_preprocessing_taileye(path):
    #gather raw data
    with os.scandir(path) as entries:
        for entry in entries:
            if 'behavior_log' in entry.name:
                bh_path = entry.path
            elif 'metadata' in entry.name:
                metadata_path = entry.path
    taileye_df = pd.read_csv(bh_path, sep=';', index_col=0)
    
    file = open(metadata_path)
    metadata = file.read()
    start_time = metadata[metadata.find('t_protocol_start": "'):].split('"')[2]
    start_time = dt.strptime(start_time, '%Y-%m-%dT%H:%M:%S.%f')
    file.close()
    
    #normalize for time
    running_time = []
    for t in taileye_df.t:
        time = start_time + timedelta(seconds = t)
        running_time = running_time + [time]
    taileye_df.loc[:, 'real_time'] = running_time
    
    #save file
    taileye_df.to_hdf(path + 'taileye_df.hdf', key = 'taileye', mode = 'w')
```
